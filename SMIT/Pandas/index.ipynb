```python
import pandas as pd
df = pd.read_csv("data.csv")
print(df.head(5))
```

```python
print(df.info())
```

```python

print(df.describe())
print("\n")
print(df.describe().T)
```

```python
print(df["Calories"].max())
print(df["Calories"].mean())
print(df["Calories"].median())
print(df["Calories"].sum())
print(f"25% of Total Calories is: {df["Calories"].sum() * 25 / 100}")
print(f"50% of Total Calories is: {df["Calories"].sum() * 50 / 100}")
```
