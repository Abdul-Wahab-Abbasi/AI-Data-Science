**A2.** Given `a = np.array([4, 8, 15, 16, 23, 42])`, compute the **sum**, **mean**, **median**, **standard deviation**, and **variance**.

```python
a = np.array([4, 8, 15, 16, 23, 42])
print(f"sum: {np.sum(a)}\nmean: {np.mean(a)}\nmedian: {np.median(a)}\nstandard deviation: {np.std(a)}\nvariance: {np.var(a)}")
```

**A3.** For a `(3, 4)` array `m = np.arange(12).reshape(3, 4)`, compute `np.sum(m, axis=0)` and `np.sum(m, axis=1)`. What is the output shape of each?

```python
m = np.arange(12).reshape(3, 4)
np.sum(m, axis=0) # will give the sum of each column
np.sum(m, axis=1) # will give the sum of each row
```

**A4.** Given `arr = np.array([3, 1, 7, 2, 9, 4])`, find: (a) index of the max, (b) index of the min, (c) the array sorted ascending.

```python
arr = np.array([3, 1, 7, 2, 9, 4])
print(f"(a) Index of the max {np.argmax(arr)}")
print(f"(b) Index of the min {np.argmin(arr)}")
print(f"(c) The array sorted ascending {np.sort(arr)}")
```

**A5.** Use `np.cumsum` and `np.diff` on `np.array([2, 5, 9, 14])`. Why does `diff` return one fewer element?

```python
arr = np.array([2, 5, 9, 14])
print(arr)
print(np.cumsum(arr))
print(np.diff(arr)) # will give one fewer element because it takes 2 adjacent elements.
```

**A6.** Reshape `np.arange(12)` into a `(3, 4)` matrix, then compute the **mean of each column** and the **max of each row**.

**A7.** From `data = np.array([12, 7, 22, 3, 18, 9])`, use **boolean masking** to (a) select all values > 10, and (b) count how many values are even.

```python
data = np.array([12, 7, 22, 3, 18, 9])
gr_10 = data[data > 10]
evens = np.sum(data % 2 == 0)
print(f"Values greater than 10: {gr_10}\nTotal even values: {evens}")
```

**A9.** Create a 3×3 identity matrix and a 3×3 matrix of all sevens without typing the numbers manually.

**A10.** Given `s = np.array([120, 90, 200, 175, 60, 145])`, compute the **total**, the **average**, and the **percentage each month** contributes to the total.

```python
s = np.array([120, 90, 200, 175, 60, 145])
total = np.sum(s)
average = np.average(s)
contributions = np.round((s / total * 100),2)
print(f"Total: {total}")
print(f"Average: {average:.2f}")
print("Percent contributions by month:")
for i, pct in enumerate(contributions, 1):
    print(f" Month {i}: {pct:.2f}%")
```

**B5.** Simulate rolling a die **1000 times**. Estimate P(rolling a 6) and compare it to 1/6.

**B6.** In a comment, explain the difference between `np.random.rand`, `np.random.randn`, and `np.random.randint`. Show one example of each.

```python
# np.random.rand generates floats from a uniform distribution over [0, 1)
print(np.random.rand(2, 3))

# np.random.randn generates floats from the standard normal distribution (mean 0, std 1)
print(np.random.randn(1))

# np.random.randint generates random integers in [low, high)
print(np.random.randint(1, 6))
```

**B8.** Generate 10,000 samples from a normal distribution with mean 50 and std 5. Report the sample mean and std.

```python
sample = np.random.normal(50, 5, 10000)
print(f"Sample mean: {sample.mean():.2f}")   # ≈ 50
print(f"Sample std : {sample.std():.2f}")    # ≈ 5
```

**B9.** Roll two dice 100,000 times. Estimate P(sum == 7).
