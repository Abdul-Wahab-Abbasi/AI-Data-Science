```python
x = [0,1,2,3,4,5]
y = [10,10,15,15,20,20]
plt.plot(x, y)
plt.title("Mera graph")
plt.ylabel("Y Lable")
plt.xlabel("X Lable")
plt.show()
```

```python
x, y = [10, 16, 20], [20, 15, 30]
x2, y2 = [8, 14, 18], [16, 11, 26]
plt.plot(x, y, 'g', label = "Line 1", linewidth=2)
plt.plot(x2, y2, 'b', label = "Line 2", linewidth=2)
plt.title("Mera graph")
plt.ylabel("Y Lable")
plt.xlabel("X Lable")
plt.grid(True, color="k")
plt.show()
```
