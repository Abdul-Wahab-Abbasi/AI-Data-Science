### Drop NA and Fill NA


```python
import pandas as pd
df = pd.read_csv('data.csv')
df.info() # before removing null rows 
newDf = df.dropna()
newDf.info() # after removing null rows 
```

```python
df.info() # before filling null rows 
newDf = df.fillna(100)
newDf.info() # after filling null rows 
```

### Changing orignal dataset with "inplace" argument

```python
df.info() # before removing null rows 
df.dropna(inplace = true)
df.info() # after removing null rows 
```

### Replace Only For Specified Columns

```python
df = pd.read_csv('data.csv')
df.info()
df.fillna({"Calories": 130}, inplace=True)
df.info()
```
